```python
import json
from pathlib import Path
import pandas as pd
```

```python
# JSON 파일이 있는 폴더
json_folder = Path('./final_metadata/')
print(Path.cwd())
```

```python
rows = []
for json_file in sorted(json_folder.glob("*.json")):
    with json_file.open("r", encoding="utf-8-sig") as file:
        data = json.load(file)

    rows.append(data)  # 모든 key와 값을 추가

if not rows:
    raise FileNotFoundError(f"JSON 파일이 없습니다: {json_folder}")

df = pd.DataFrame(rows)

output_file = json_folder / "전체_JSON_데이터.xlsx"
df.to_excel(output_file, index=False)

print(f"{len(df)}개 저장 완료: {output_file}")
```
